```
# practice 1
# Initial list of numbers
numbers = [87, 63, 68, 73, 32, 66, 84, 94, 96, 83]

# 1. Create a 10x10 diagonal matrix using a for-loop
A = np.zeros((10, 10), dtype=int)
for i in range(len(numbers)):
    A[i, i] = numbers[i]

# 2. Compute the transpose and matrix product
A_T = A.T
B = np.dot(A, A_T)  # equivalent to A @ A_T

# 3. Compute the trace of the product matrix
trace_result = int(np.trace(B))

print("Trace of the product matrix:", trace_result)
```

```
# practice 1-instead

numbers = [87, 63, 68, 73, 32, 66, 84, 94, 96, 83]

# Alternative method: Create the diagonal matrix directly using np.diag
A_alt = np.diag(numbers)

# Compute the trace of A @ A.T to verify
trace_alt = int(np.trace(A_alt @ A_alt.T))

print("Trace using alternative method:", trace_alt)
```

```
# practice 2
# 1. Create a 10x10 matrix with all initial values equal to 1
MA = np.ones((10, 10), dtype=int)

# Dictionary of numbers and their (row, col) positions
numbers_positions = {
    98: (0, 1),
    62: (2, 2),
    87: (3, 4),
    35: (5, 1),
    4: (6, 3),
    24: (7, 0),
    28: (1, 5),
    91: (4, 4),
    49: (8, 2),
    100: (9, 9)
}

# 2. Update matrix MA using the dictionary
for val, (r, c) in numbers_positions.items():
    MA[r, c] = val

# 3. Extract submatrix MB (rows 3 to 6, cols 2 to 5)
MB = MA[3:7, 2:6]

# 4. Perform multiplication and sum operations
# Result 1: Multiply all elements in MB
result_1 = np.prod(MB)

# Result 2: Add 10 to each element in MB (matrix MC) and calculate the sum
MC = MB + 10
result_2 = np.sum(MC)

# 5. Calculate the final result
final_result = int(result_1 + result_2)

print("Result 1:", result_1)
print("Result 2:", result_2)
print("Final Result:", final_result)
```

```
import pandas as pd

# ==========================================
# Step 1: Generate images.csv
# ==========================================

# Define file names (first 5 ascending, next 5 in reverse order)
first_five = ['img_1.jpg', 'img_2.jpg', 'img_3.jpg', 'img_4.jpg', 'img_5.jpg']
file_names = first_five + first_five[::-1]

# Generate values starting from 32, increasing by 2 for each line
values = [32 + 2 * i for i in range(10)]

# Create DataFrame
df = pd.DataFrame({
    'file_name': file_names,
    'value': values
})

# Save to CSV
csv_filename = 'images.csv'
df.to_csv(csv_filename, index=False)
print("CSV file generated successfully.")


# ==========================================
# Step 2: Read CSV and sum specific rows
# ==========================================

# Read CSV back into Python
df_read = pd.read_csv(csv_filename)

# Extract Row 2, Row 7, and Row 8 (0-indexed: index 1, index 6, index 7)
val_row2 = df_read.loc[1, 'value']  # 2nd row -> value: 34
val_row7 = df_read.loc[6, 'value']  # 7th row -> value: 44
val_row8 = df_read.loc[7, 'value']  # 8th row -> value: 46

# Calculate the sum
total_sum = int(val_row2 + val_row7 + val_row8)

print(f"Row 2 value: {val_row2}")
print(f"Row 7 value: {val_row7}")
print(f"Row 8 value: {val_row8}")
print(f"Final Answer (Sum): {total_sum}")
```
